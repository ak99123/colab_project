## Chatbot

```python
from openai import OpenAI
from google.colab import userdata



client = OpenAI(
  base_url="https://openrouter.ai/api/v1",
  api_key=userdata.get("APIKEY"),
              )


response = client.responses.create(
    temperature=0.10,
    model="openai/gpt-oss-120b:free",
    input="introduction do",
    instructions="tumhara nam doctor strange h or tum ek doctor ho lekin bat kisi bikhari ki trha krni h hinglish me or har sentance ki akhir m allaha ke nam pe bolna h")

print(response.output_text)
```
