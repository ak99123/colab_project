## Chatbot

```python
!pip install -q openai

from openai import OpenAI
from google.colab import userdata

client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=userdata.get("APIKEY")
)

MODEL = "openai/gpt-3.5-turbo"


PROMPT_TEMPLATE = """
You are a professional car mechanic.
Role: You are an expert who summarizer every problem.

Task: find the given problem in car clearly and accurately.

Dynamic Variables:
- Language: {language}
- Tone: {tone}
- Summary length: {summary_length}
- Input text: {input_text}

Instructions:
1. Read the input text carefully.
2. Keep the meaning correct.
3. Do not add fake information.
4. Use a {tone} tone.
5. Write the answer in {language}.
6. Make the summary {summary_length}.

Output Format:
Title:
Key Points:
1.
2.
3.
Conclusion:
"""

def create_prompt(input_text, language="Hinglish", tone="friendly nighbourhood", summary_length="short"):
    prompt = PROMPT_TEMPLATE.format(
        input_text=input_text,
        language=language,
        tone=tone,
        summary_length=summary_length
    )
    return prompt

def generate_summary(input_text, language="Hinglish", tone="friendly nighbourhood", summary_length="short"):
    final_prompt = create_prompt(
        input_text=input_text,
        language=language,
        tone=tone,
        summary_length=summary_length
    )

    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {
                "role": "system",
                "content": "You are a professional car mechanic. You help clients in English, Urdu, and Roman Urdu."
            },
            {
                "role": "user",
                "content": final_prompt
            }
        ],
        temperature=0.4
    )

    return response.choices[0].message.content


text_1 = """
meri car me masla h garmi m ac off krta hu to garmi hoti h
and switch off krta hu to car band hojati h
lights off krta hu to car ki heaflights off hojati h Q
akhit mere sath aesa Q ho rha h
"""

print("========== TEST 1 ==========")
print(generate_summary(
    input_text=text_1,
    language="roamn urdu",
    tone="ek dam munna bahi MBBS waly charater ke andaz me",
    summary_length="detailed"
))
```
